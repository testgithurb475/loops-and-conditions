## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
# 2. BMI Health Checker
weight = float(input("Enter weight: "))
height = float(input("Enter height: "))
bmi = weight / (height * height)

if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi <= 24.9:
    category = "Normal"
elif 25 <= bmi <= 29.9:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI = {bmi:.2f} → {category}")
```

## 🌡️ 8. Temperature Unit Converter

### Goal:
    Build a small converter for a weather app.

### Steps:

    Show a menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius.
    
    Ask for the choice and then the temperature.
    
    Choice 1 → F = C * 9/5 + 32.
    
    Choice 2 → C = (F - 32) * 5/9.
    
    Any other choice → "Invalid choice".

### Example:

    Enter choice: 1
    Enter temperature: 100
    212.0 F

```python
# 8. Temperature Unit Converter
choice = int(input("Enter choice: "))
temp = float(input("Enter temperature: "))

if choice == 1:
    f = temp * 9 / 5 + 32
    print(f"{f:.1f} F")
elif choice == 2:
    c = (temp - 32) * 5 / 9
    print(f"{c:.1f} C")
else:
    print("Invalid choice")
```

### Goal:
    Help a shopper find the lowest price of the same product from three shops.

    Use nested if statements to find the smallest price (do not use min()).
    
    Print the lowest price and the shop number.

### Example:

    Enter prices: 1450, 1200, 1380
    Lowest price: 1200 (Shop 2)

```python
# 9. Cheapest Price Finder
p1 = float(input("Enter price in Shop 1: "))
p2 = float(input("Enter price in Shop 2: "))
p3 = float(input("Enter price in Shop 3: "))

if p1 < p2:
    if p1 < p3:
        lowest = p1
        shop = 1
    else:
        lowest = p3
        shop = 3
else:
    if p2 < p3:
        lowest = p2
        shop = 2
    else:
        lowest = p3
        shop = 3

print(f"Lowest price: {lowest:.0f} (Shop {shop})")
```

## 📐 20. Roof Truss Angle Check

### Goal:
    Help a civil engineer check whether three angles can form a triangle.

### Steps:

    Ask for three angles.
    
    Valid only if every angle is greater than 0 AND the sum of the three angles is exactly 180.
    
    Print "Valid triangle" or "Invalid triangle".

### Example:

    Enter angles: 60, 60, 60
    Valid triangle

```python
# 20. Roof Truss Angle Check
angles_str = input("Enter angles: ")
angles = [float(a.strip()) for a in angles_str.split(',')]

if len(angles) == 3:
    a1, a2, a3 = angles
    if a1 > 0 and a2 > 0 and a3 > 0 and (a1 + a2 + a3 == 180):
        print("Valid triangle")
    else:
        print("Invalid triangle")
else:
    print("Invalid input")
```

## 🧱 47. Largest Square Tile

### Goal:
    A floor is 48 ft by 18 ft. Find the biggest square tile that fits without cutting.

### Steps:

    Ask for the length and the width.
    
    Use a while loop with Euclid's method: while b is not 0, set a, b = b, a % b.
    
    The answer is the final value of a (the GCD).
    
    Print the tile size.

### Example:

    Length: 48
    Width: 18
    Largest tile = 6 ft

```python
# 47. Largest Square Tile
length = int(input("Length: "))
width = int(input("Width: "))

a, b = length, width
while b != 0:
    a, b = b, a % b

print(f"Largest tile = {a} ft")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

```python
# 50. Weekly Temperature Report
temps = [30, 32, 28, 35, 31, 29, 36]

total = 0
for t in temps:
    total += t
avg = total / len(temps)

count = 0
for t in temps:
    if t > avg:
        count += 1

print(f"Average = {avg:.2f}")
print(f"Days above average = {count}")
```
