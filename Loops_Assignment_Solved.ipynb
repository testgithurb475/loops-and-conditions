## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []
for _ in range(5):
    val = int(input("Enter marks: "))
    if val == -1:
        print("stop")
        break
    if val < 0 or val > 100:
        print("Invalid marks")
        continue
    marks.append(val)

if len(marks) > 0:
    avg = sum(marks) / len(marks)
    print("Valid Marks:", marks)
    print("Average =", avg)
else:
    print("No valid marks entered.")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
account_created = False

while True:
    print("\n--- Banking Menu ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    
    choice = input("Enter choice (1-5): ").strip()
    
    if choice == '1':
        name = input("Enter account holder name: ")
        account_created = True
        print(f"Account created successfully for {name}!")
    elif choice == '2':
        if not account_created:
            print("Please create an account first!")
            continue
        amt = float(input("Enter deposit amount: "))
        if amt > 0:
            balance += amt
            print(f"Deposited {amt}. New Balance = {balance}")
        else:
            print("Invalid deposit amount!")
    elif choice == '3':
        if not account_created:
            print("Please create an account first!")
            continue
        amt = float(input("Enter withdrawal amount: "))
        if 0 < amt <= balance:
            balance -= amt
            print(f"Withdrawn {amt}. Remaining Balance = {balance}")
        else:
            print("Insufficient funds or invalid amount!")
    elif choice == '4':
        if not account_created:
            print("Please create an account first!")
            continue
        print(f"Current Balance = {balance}")
    elif choice == '5':
        print("Exiting Banking System. Thank you!")
        break
    else:
        print("Invalid choice, try again.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for s in range(1, 6):
    print(f"Student {s}:")
    marks = []
    failed = False
    fail_sub = 0
    for sub in range(1, 5):
        m = float(input(f"Subject {sub} marks: "))
        marks.append(m)
        if m < 33:
            failed = True
            fail_sub = sub
    
    total = sum(marks)
    avg = total / 4
    if failed:
        print(f"Student {s}: {marks} → Failed in Subject {fail_sub}")
    else:
        if avg >= 80:
            grade = "A"
        elif avg >= 60:
            grade = "B"
        else:
            grade = "C"
        print(f"Student {s}: {marks} → Grade: {grade}")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
for pkg in range(1, 4):
    ratings = []
    print(f"Package {pkg}:")
    for feat in range(1, 4):
        r = float(input(f"Rating for feature {feat} (0-5): "))
        if r == 0:
            continue
        ratings.append(r)
    
    if len(ratings) > 0:
        avg = sum(ratings) / len(ratings)
        print(f"Package {pkg}: {ratings} → Avg = {avg:.2f}")
    else:
        print(f"Package {pkg}: No valid ratings.")
```
